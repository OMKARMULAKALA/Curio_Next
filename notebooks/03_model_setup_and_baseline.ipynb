# Model Setup and Baseline

Load Qwen2-Audio and run a small baseline inference sample. This is not the main result.

```python
import yaml
from src.data.dataset_loader import load_split
from src.models.qwen2_audio import load_qwen2_audio, generate_answer
from src.utils.audio import load_audio

model_cfg = yaml.safe_load(open('configs/model.yaml', encoding='utf-8'))
record = load_split('validation')[0]
record.to_dict()
```

```python
model, processor = load_qwen2_audio(model_cfg['model_name'], **model_cfg.get('from_pretrained_kwargs', {}))
```

```python
audio, sr = load_audio(record.audio_path)
prediction = generate_answer(model, processor, audio, record.question, record.choices, sampling_rate=sr)
{'prediction': prediction, 'expected': record.answer}
```
