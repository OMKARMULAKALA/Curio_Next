# Error Analysis

Inspect incorrect predictions and summarize failures by question type.

```python
import json
from pathlib import Path
from src.evaluation.error_analysis import identify_errors, summarize_errors

path = Path('results/predictions/predictions.jsonl')
records = [json.loads(line) for line in path.open(encoding='utf-8')] if path.exists() else []
len(records)
```

```python
errors = identify_errors(records)
summarize_errors(errors) if errors else {'message': 'No prediction errors available yet.'}
```

```python
import pandas as pd

df = pd.DataFrame(errors)
df.head() if not df.empty else df
```

```python
if not df.empty and 'question_type' in df:
    df.groupby('question_type').size().sort_values(ascending=False)
else:
    print('No errors to group yet.')
```
